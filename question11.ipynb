```python
def compute_mean(a_list):
    total = 0
    count = 0
    for number in a_list:
        total = total + number
        count = count + 1
    average = total / count
    print(average)

compute_mean([2, 4, 6, 8])
```

This function takes a list of numbers. It loops through the list, adding each value to a running total and counting how many values there are. It then divides the total by the count to get the mean and prints the result.

$$\bar{x} = \frac{x_1 + x_2 + \cdots + x_n}{n} = \frac{1}{n}\left(\sum_{i=1}^{n} x_i\right)$$
